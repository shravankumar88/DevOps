### List
- List is a mutable(changable example add using apppend), ordered sequence
- Elemments can be differnt datatypes
- Lists are defined using square brackets []
- Duplicates are allowed
- Exampe: My_list = [2,3.5,"Learn",True]
- Even a single element the type will be List ex: data=[1] print(type(data))

### Tuple
- Tuple is a immmutable(unchangle), ordered sequence of elements
- Similar to lists can`t be modified once created 
- Tuples are defined using parenythesis()
- Elememets can be differnt datatypes
- Duplicates are allowed
- Tuple single element will be int we should use , to consider tuple while using type
- Example: My_tuple = (2,3,3.5,"str")

### Set
- Set is an unordered collection of unique elememts(duplicates not allowed)
- Set is a mutable data type(changable)
- It allows only elemtnts as immumtable datatype
    - N(Int,Float)T(Tuple)S(STRING)
- Sets are defined using curly brackets {}
- Example: My_set = {2,3,3.5,"str"}

### Dictionary
- Stores data in Key-value pair. We can use the key to get its value. Can be used for storing related information like a user`s profile or product details
- It an ordered collection of key value pairs and mutable
- Key must be unique within a dictinoary
- Defined usinf curly braces {} and colon :
- Key should be immutable
- Example : my_dictinoary = {"name" : "Python", "Class" : 45}
