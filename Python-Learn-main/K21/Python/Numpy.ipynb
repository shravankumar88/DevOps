
> ## NumPy - Numerical PYthon  
1. It's a library consisting of multidimensional array objects.

---

> ## NumPy vs List

| Feature                | Python List                                        | NumPy Array                                                |
|------------------------|----------------------------------------------------|------------------------------------------------------------|
| **Data Type Support**  | Able to store different data types in the same list| Can store only one data type in an array at any time       |
| **Memory Allocation**  | Items stored in random memory locations            | Items stored sequentially in memory                        |
| **Use Case**           | Good for scenarios where list can grow dynamically | Good for scenarios with fixed-size, same data type elements|
| **Library Requirement**| Inbuilt data type                                  | Requires external NumPy library                            |
| **Functionality**      | Has more inbuilt functions                         | Lightweight with no extra functions (less memory usage)    |

> ## Array Shapes

1. `shape(4,)`       → 1D array with 4 columns  
2. `shape(2, 3)`     → 2 rows and 3 columns  
3. `shape(4, 3, 2)`  → 4 layers, 3 rows, 2 columns  

---

> ## Key NumPy Attributes & Methods

| No. | Attribute / Method     | Description |
|-----|------------------------|-------------|
| 1.  | `.ndim`                | Check the number of dimensions |
| 2.  | `.shape`               | Print the shape of the array |
| 3.  | `.dtype`               | Check the data type of the array |
| 4.  | `type()`               | Check the type of the array (e.g. `<class 'numpy.ndarray'>`) |
| 5.  | `len()`                | Length of the array |
| 6.  | `np.arange(start, stop, step)` | Like Python `range()` but returns an array |
| 7.  | `np.eye(n)`            | Returns a 2D array with 1s on the diagonal, 0s elsewhere |
| 8.  | `np.zeros`, `np.ones`, `np.full` | Generate arrays filled with specified values |
| 9.  | `np.diag([values])`    | Creates a diagonal matrix |
| 10. | `.reshape()`           | Change the shape without changing data |
|     |                        | Example reshapes for 12 elements: `(3,4)`, `(4,3)`, `(2,6)`, `(6,2)`, `(2,3,2)` |
| 11. | `np.ravel()`           | Flattens n-dimensional array to 1D (returns view if possible) |
| 12. | `flatten()`            | Flattens n-dimensional array to 1D (always returns a copy) |

---

> ## 🔁 Ravel vs Flatten

- `ravel()` → Changes **may** affect original (returns **view**)
- `flatten()` → Changes **do not** affect original (returns **copy**)

---

> ## Indexing and Slicing

- Slicing syntax: `[row_start:row_end, col_start:col_end]`
- 3D Slicing: `[layer_start:layer_end, row_start:row_end, col_start:col_end]`

---

> ## Copy vs View

| Concept      | Description |
|--------------|-------------|
| `.copy()`    | Creates a new array in a **different memory location**. Changes do **not** affect original. *(Like `flatten`)* |
| `.view()`    | Creates a new array **sharing the same memory**. Changes **do** affect original. *(Like `ravel`)* |
| `np.shares_memory(a, b)` | Returns `True` if two arrays share the same memory |

### why lists are most commonly used to create NumPy arrays.
1. Lists are ordered and structured
  Lists maintain order and allow nesting (for 2D, 3D arrays).

  You can do things like:

  [[1, 2], [3, 4]]  # 2D list
  [[[1], [2]], [[3], [4]]]  # 3D list
  This structure maps perfectly to how NumPy organizes multi-dimensional arrays.

2. Lists support mixed and uniform data
  While Python lists allow mixed types, we usually use them with consistent types for NumPy.

  NumPy automatically infers a consistent dtype from the list.

  Example:

  import numpy as np
  np.array([1, 2, 3])       # → int array
  np.array([1.0, 2.0, 3.0]) # → float array

3. Easy to write and read
  Lists are:

  Simple to type: [1, 2, 3]

  Easy to visualize, especially when nested

  Familiar to most Python users

  This makes lists the default choice when starting with NumPy arrays.

  Why not use other types as often?
  >Type	     Why it's not ideal for NumPy arrays.

  >Tuple	   Immutable, but okay — just less common

  >Set	     Unordered and no duplicates — not suitable for structured arrays

  >Dict	     Key-value format doesn’t map well to arrays

  >String	   Treated as a sequence of characters — only useful in certain cases

  Bottom line:
  Lists are the most natural and flexible way to create arrays — especially for multidimensional data — and that's why they're used most often.

### reshape, ravel,flatten
- .reshape - changing the shape (dimensions) of an array without changing its data.
  - Example: An array range (1,13).To reshape what will be possibilities for 12
  - 12 we can reshape into (3,4)(4,3)(1,12)(12,1)(2,6)(6,2)(2,3,2)...
- np.ravel() - it converts ndimension into 1D
- np.ravel() - it converts ndimension into 1D
>Differnce b/n ravel and faltten

Changes to ravel() output may affect the original array (if it's a view)	Changes to flatten() output do not affect the original array

````python
#.reshape - Reshaping means changing the shape of a array(12)
reshape_ex = np.arange(1,13)
print(reshape_ex)
print(reshape_ex.shape)
print(reshape_ex.ndim)
````

````python
reshape_ex_1 = reshape_ex.reshape((4,3))
print(reshape_ex_1.shape)
print(reshape_ex_1.ndim)
print(f"reshape_ex_1 \n {reshape_ex_1} \n { '-' * 40}")
reshape_ex_2 = reshape_ex.reshape((3,4))
print(reshape_ex_2.shape)
print(reshape_ex_2.ndim)
print(f"reshape_ex_2 \n {reshape_ex_2} \n { '-' * 40}")
reshape_ex_3 = reshape_ex.reshape((2,3,2))
print(reshape_ex_3.shape)
print(reshape_ex_3.ndim)
print(f"reshape_ex_3 \n {reshape_ex_3} \n { '-' * 40}")
````

````python
#np.ravel() - it converts ndimension into 1D
ravel_ex = np.ravel(reshape_ex_3)
print(ravel_ex)
print(ravel_ex.ndim)
print(ravel_ex.shape)

````

````python
#flatten(Converst array into 1D)
flatten_ex = reshape_ex_3.flatten()
print(flatten_ex)
print(flatten_ex.ndim)
print(flatten_ex.shape)
````

````python
diff_rav_fal = np.array([[3,2],[4,5]])
print(diff_rav_fal.ndim)
print(diff_rav_fal.shape)
print(f"diff_rav_fal \n {diff_rav_fal} \n { '-' * 40}")
diff_rav= np.ravel(diff_rav_fal)
print(diff_rav.ndim)
print(diff_rav.shape)
print(f"diff_rav \n {diff_rav} \n { '-' * 40}")
diff_fal = diff_rav_fal.flatten()
print(diff_fal.ndim)
print(diff_fal.shape)
print(f"diff_fal \n {diff_fal} \n { '-' * 40}")
#changing the array
diff_rav_fal[0,0] = 100
print(f"diff_rav_fal \n {diff_rav_fal} \n { '-' * 40}")
print(f"diff_rav \n {diff_rav} \n { '-' * 40}")
print(f"diff_fal \n {diff_fal} \n { '-' * 40}")
````

### NumPy Copy vs View
- .shares_memory() - Use np.shares_memory(a, b) to check whether two NumPy arrays share the same memory. It returns True if they do, and False otherwise.
- .view - The view() function creates a new array that shares the same memory, so changes in one array affect the other like ***flattern***.
- .copy() - The copy() function creates a new array with a different memory location, so changes in the copied array do not affect the original like ***ravel***

````python
x1 = np.arange(11,55,4)
print(x1)
````

````python
#assign
x2 = x1 #view
print(x1)
print(x2)
````

````python
#change the element of copy_vs_view_1 as 10
x2[0] = 100
print(x1)
print(x2)
````

````python
print(id(x1))
print(id(x2))
````

````python
#shares_memory()
# To check the variables whether share the memory or not
print(np.shares_memory(x1,x2))
````

````python
#.view function
print(x2)
x3 = x2.view()
print(x3)
view_share_memory = np.shares_memory(x2,x3)
print(f"view_share_memory is : {view_share_memory}") # it shared same memory
````

````python
#.copy() - It will copy the variable. But, the memeory location will be differnt.
x4 = x1.copy()
print(x1)
print(x4)
print(id(x1)) # the memory location is different
print(id(x4))
````

````python
 # if you create a copy of an array using the copy() function, changing an element in the copied array does not affect the original array, because they are stored in different memory locations.
x4[0]= 999
print(x4)
print(x1)
````

````python
print(np.shares_memory(x1,x4))
````

### Adding, Insert and Delete NumPy Array
- np.insert - To insert the elements accroding to the index.
- np.delete - Delete the elementts
- np.append - Elements will be add at last
  - in 2D (rows,columns) ***INSERT/DELETE ARRAY***
    - axis = 0 Add/Removes in rows
    - axis = 1 Add/Removes in column
  - in 3D (layers,rows,column) ***INSERT/DELETE ARRAY***
    - axis = 0 Add/Removes layers
    - axis = 1 Add/Removes row in each layer
    - axis = 2 Add/Removes column in each layer
- np.concatenate - Most flexible and general-puporse. We must manully specify the axis to join. Doesn`t not add a new axis (unlike stack)

# 📚 Understanding Axis in 3D NumPy Arrays

A 3D NumPy array is shaped as **(D, R, C)**:
- `D`: Depth (number of slices or layers)
- `R`: Rows
- `C`: Columns

---

## 🔄 Axis Meaning in Aggregation Functions

| Axis    | Affects             | Collapses Along     |
|---------|----------------------|----------------------|
| `axis=0` | Between **slices**    | Collapses **depth**  |
| `axis=1` | Between **rows**      | Collapses **rows**   |
| `axis=2` | Between **columns**   | Collapses **columns**|

---

## 🧠 Axis Representation in 3D Arrays

| Axis    | Represents             | Example Term        |
|---------|------------------------|---------------------|
| `0`     | Layers                 | Slices / Depth      |
| `1`     | Rows per layer         | Horizontal lines    |
| `2`     | Columns per row        | Vertical lines      |

---

## 🔁 In Aggregate Functions (for 3D Arrays):

- `axis=0` → Layer-wise operation (collapses depth)
- `axis=1` → Column-wise across each layer
- `axis=2` → Row-wise across each layer

---

### 📦 Visualization

> A **3D array** is like a **stack of 2D sheets**  
> Each sheet = one layer (**axis 0**)  
> Inside each sheet:
> - Rows = **axis 1**
> - Columns = **axis 2**


# 📘 Detailed Explanation of `np.std` on a 3D Array

## 🔢 Input Code

```python
import numpy as np

std_3d = np.arange(1, 9).reshape(2, 2, 2)
print(np.std(std_3d, axis=0))
print(np.std(std_3d, axis=1))
print(np.std(std_3d, axis=2))
```

---

## 📐 3D Array Structure

The array looks like this:

```
std_3d =
[[[1, 2],
  [3, 4]],

 [[5, 6],
  [7, 8]]]
```

Shape: (2, 2, 2) → (layers, rows, columns)

---

## 📏 Standard Deviation Formula

\[
\sigma = \sqrt{ \frac{1}{N} \sum_{i=1}^{N} (x_i - \mu)^2 }
\]

Where:  
- \( \mu \) = Mean  
- \( N \) = Number of elements  
- \( x_i \) = Each element

---

## ✅ `np.std(std_3d, axis=0)` → Layer-wise Comparison

You're comparing the two layers **element by element**:

```
Element-wise pairing:
[[1, 2],   [[5, 6]] → compare: (1,5), (2,6), (3,7), (4,8)
 [3, 4]]

Mean for each pair:
- (1,5): μ = 3 → σ = √[(1-3)² + (5-3)²]/2 = √[4 + 4]/2 = √4 = 2
- (2,6): μ = 4 → σ = 2
- (3,7): μ = 5 → σ = 2
- (4,8): μ = 6 → σ = 2

```python
np.std(std_3d, axis=0)
# Output:
[[2. 2.]
 [2. 2.]]
```

---

## ✅ `np.std(std_3d, axis=1)` → Row-wise Within Each Layer

Now comparing **rows inside each layer**.

### Layer 1:
```
[[1, 2],
 [3, 4]]
```
Element-wise across rows:
- (1,3): μ = 2 → σ = √[(1-2)² + (3-2)²]/2 = √1 = 1
- (2,4): μ = 3 → σ = √[(2-3)² + (4-3)²]/2 = √1 = 1

### Layer 2:
```
[[5, 6],
 [7, 8]]
```
- (5,7): μ = 6 → σ = 1
- (6,8): μ = 7 → σ = 1

```python
np.std(std_3d, axis=1)
# Output:
[[1. 1.]
 [1. 1.]]
```

---

## ✅ `np.std(std_3d, axis=2)` → Column-wise (Last Dimension)

Now computing std **within each row**.

### Layer 1:
- [1, 2]: μ = 1.5 → σ = √[(1-1.5)² + (2-1.5)²]/2 = √0.25 = 0.5
- [3, 4]: μ = 3.5 → σ = 0.5

### Layer 2:
- [5, 6]: μ = 5.5 → σ = 0.5
- [7, 8]: μ = 7.5 → σ = 0.5

```python
np.std(std_3d, axis=2)
# Output:
[[0.5 0.5]
 [0.5 0.5]]
```

---

## 📊 Summary Table

| Axis | Meaning      | Operates On                | Output Shape | Example Result           |
|------|--------------|----------------------------|---------------|---------------------------|
| 0    | Layer-wise   | Between blocks (depth)     | (2, 2)        | `[[2. 2.], [2. 2.]]`      |
| 1    | Row-wise     | Across rows in each layer  | (2, 2)        | `[[1. 1.], [1. 1.]]`      |
| 2    | Column-wise  | Within each row            | (2, 2)        | `[[0.5 0.5], [0.5 0.5]]`  |
