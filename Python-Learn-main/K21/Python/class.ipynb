```python
#range(start,stop,step)
range(5)
print(range(5,10,2))
for i in range(5):
    print(i)
for i in range(1,6):
    print (i)
for i in range(1,10,2):
    print (i)
for i in range(10,1,-2):
    print (i)
```

```python
#While Loop The loop contuines to execute as long as condition is TRUE
count = 0
while count<5:
    print(count)
    count=count+1
```

```python
#BREAK -  The break stts exists the loop permaturely
for i in range(10):
    if i==5:
        break
    print(i)
```

```python
#Continue -skips the current iteration and countine with the next.
for i in range(10):
    if i%2 == 0:
        continue
    print(i)
```

```python
# pass
for i in range(5):
    if i ==3:
        pass
    print(i)
```

```python
#Nested Loops
for i in range(3):
    for j in range(2):
        print(i*j)
```
