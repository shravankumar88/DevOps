Functions
- It is a block of code that performs a specfic task. Help in organizing code,resuing code and improve readbaility 
- Syntax
    - def function_name(parameters):
    - """Doc string need base to under about the function"""
    - Function body or logic 
    - return expression
- it supports mutiple parameter and we can define default parameters
- Suppose instaed of displaying the aoutput using print we can store using return expression
- if we don`t use the return it will consider as NONE and we can pass differnt data types like return [a*b],a
- Return will store the values and we can reuse it by assign a varibale to the funtion and perform the action

SCOPE
 - Global Scope - define a variable outside of the function
 - Local Scope - define a variable inside the function
 - if we define varaiable global and local inside the function it consider local

 Expection Handling
 - try - experimental code will be written in try
 - expect - in try if you expect and errors that can be defined in expect
 - else - if try executed succesfully else will be executed like define print
 - finally - it executes on every step and display the output respective of success or failed conditions in try,expect, else

| Feature        | `print()`                              | `return`                                    |
| -------------- | -------------------------------------- | ------------------------------------------- |
| Purpose        | Displays output to console             | Sends value back to caller                  |
| Output Usage   | Used for debugging or showing messages | Used for storing or further computation     |
| Executes where | Inside the function                    | Inside the function but used outside too    |
| Example        | `def show(): print("Hello")`           | `def get_msg(): return "Hello"`             |
| Calling Output | `show()` → prints `Hello`              | `msg = get_msg()` → stores `Hello` in `msg` |

```python
try:
    numb = int(input("Enter the name"))
    results = 10/numb
except ValueError:
    print("That` not a vaild number")
except ZeroDivisionError as ze:
    print(ze)
else:
    print(f"Value of number:{numb}")
finally:
    print("Succefully executed")
```

```python
def divide(a,b):
    try:
        divide=(a/b)
    except ZeroDivisionError as ze:
        print(ze)
    else:
        print(f"Value of number:{divide}")
    finally:
        return divide
```

```python
divide(10,5)
```
