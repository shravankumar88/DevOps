OOPS - Object oriented programming 
- it contains class, objects, attributes, methods
- Class - The Blueprint
    - A class like a blueprint or plan for making something.
        - ex: Imaagine you have a pplan to make robot
- Object - The Real Toy
    - An object is the real thing made from the class
    - Ex: It's like building Robot from the blueprint
- Attribute - The RObot`s feature
    - An attribute is like a feature of the robot - it`s size color or name
- Methods - The Robot`s action
    - A method is someting the robot can do like taking, waking or dancing.
self 
- what ever instance you are creating it is telling the it is belong to the current object
- (__int__) the construtor will be invoked when we define the object of the class


| Concept       | What It Means (Kid-Friendly)          | Example                   |
| ------------- | ------------------------------------- | ------------------------- |
| **Class**     | A blueprint to make a student         | `class Student1:`         |
| **Object**    | A real student created from the class | `student = Student1()`    |
| **Attribute** | Student's features like name and age  | `self.name`, `self.age`   |
| **Method**    | What the student can do               | `info()`, `eligibility()` |


```python
class Student:
    def info(self,name,age):
        self.name=name
        self.age=age
        print(f"The name of the student is {name} and age is{age} ")
    def rollnumber(self,rollnumber):
        self.rollnumber = int(rollnumber)
        print((f"the roll number of the student {rollnumber}"))
```

```python
Student_details=Student()
```
