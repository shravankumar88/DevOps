OOPS - Object Orieneted Programming System
>Types of OOPS
1. Class
2. Objects
3. Inheritance
4. Polymorphism
5. Encapsulation
6. Abstraction

1. A class is a blueprint used to create objects.
2. An object is an instance of a class.
3. Each object contains properties (to define characteristics) and functionalities (to define behavior).
4. Properties are also known as attributes or data.
5. Functionalities are also known as methods(actions objects can perform), code, or member functions.

>Class
1. Defined with keyword class. It is a user defined data structure that binds the attributes and methods.
2. Syntax class className: or class className(): best to use () while using inheritance 
>Object
1. It is collection of arributes and methods that act on thouse data
2. The process of creating an object can be called INSTANTIATION.
3. Memory is alllocated when the object is created.
>Self Keyword
1. The self parameter is used to access current class Methods and Variables
2. Without self, object-oriented programming (OOP) in Python doesn’t work as expected. This is because self connects the method to the specific object it is called on. That’s why it is essential when writing OOP programs in Python.
3. self refers to the current object.
4. It helps the program understand which object’s data to use or update.
>__init__
1. Is used to initalize the attributes. It is called a constructor



```python
class School():
    name = "Mahesh"  #attributes
    RollNo = 25
    Location = "Hyd"
    def studying(self): #methods
        print("Student is studying in the class")
    def playing(self): # without self(Error):School.studying() takes 0 positional arguments but 1 was given
        print("student is playing in school")
#Syntax: objectname=classname() for object creation
student = School() #object creation
#syntax: Objectname.variable
print(student.name)
print(student.RollNo)
print(student.Location)
#Syntax:objectname.method
student.studying()
student.playing()
```

```python
class car():
    name = "kia" #attributes
    wheels = 4 #class Variable
    def driving(self): #method
        print(f"this {self.name} car has {self.wheels} wheels")
model_car = car() 
print(model_car.name)
print(model_car.wheels)
model_car.driving()
```

```python
class School():
    def display_info(self):
        print(f"Name:{self.name}")
    def promote(self,rank):
        print(f"{self.name} has been promted to {self.grade} with {rank}")
student1=School()
student1.name="Mahesh"
student1.rollno=67
student1.grade="A"
student1.display_info()
student1.promote(56789)

#Creating student2 Object
student2=School()
student2.name="Reakesh"
student2.rollno=69
student2.grade="B"
student2.display_info()
student2.promote(56786)
```

```python
class Mobile():
    def __init__(self,brand,model,price):
        self.a=brand     #instance variable
        self.model=model
        self.price=price
    def display_details(self):
        print(f"mobile brand {self.a}")
        print(f"mobile model {self.model}")
        print(f"mobile price {self.price}")
apple=Mobile("Apple",17,"$1299")
apple.display_details()
samsung=Mobile("samsung","note 10","$1099")
samsung.display_details()
```

```python
# 125.Create a class called Calculator with two numbers as attributes.
# Use __init__ to set these numbers.
# Add methods to:
# Add the numbers
# Subtract them
# Multiply them
# Divide them

class Calculator():
    def __init__(self,number1,number2):
        self.num1=number1
        self.num2=number2
    def add(self):
        print(f"add {self.num1+self.num2}")
    def sub(self):
        print(f"sub {self.num1-self.num2}")
    def mul(self):
        print(f"mul {self.num1*self.num2}")
    def div(self):
        print(f"div {self.num1%self.num2}")
task125 = Calculator(2,3)
task125.add()
task125.sub()
task125.mul()
task125.div()
```

```python
# 126. Create a class Rectangle with width and height as attributes.
# Add methods to calculate:
# Area (width * height)
# Perimeter (2 * (width + height))
# Take input from the user to create a rectangle and print area & perimeter.
class Rectangle():
    def __init__(self,width,height):
        self.width=width
        self.height=height
    def Area(self):
        print(f"Area : {self.width*self.height}")
    def Perimeter(self):
        print(f"Perimeter:{2*(self.width+self.height)}")
task_126 = Rectangle(2,3)
task_126.Area()
task_126.Perimeter()
```

```python
# 128. Create a class Circle with:
# Attribute: radius
# Method to calculate:
# Area (πr²)
# Circumference (2πr)
# Note: Use math.pi and display the results for a given radius.
import math
class Circle():
    radius = math.pi
    def Area(self):
        print(math.pi * self.radius **2)
    def Circumference(self):
        print(2*math.pi*self.radius)
task_128=Circle()
task_128.Area()
task_128.Circumference()
```
