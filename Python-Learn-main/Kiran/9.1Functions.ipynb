Function
- A block of code which only runs when it is called.
- starts with def and functionname()
- parameters --> Function Defintion --> Formal Parameters
- Arguments  --> Function Calling   --> Actual Parameters
- Return is used to exit the function and return  a value to the caller
- collections of funtion called MODULES
Types Of Functions
- Built-in Functions:
    - Predefined funcations provided by python(input,type,len,id....)
- User-defined Functions:
    - Functions created by useres to perorm particular tasks.
    - These are classified into four types:
        1. Without arguments and without return type
        2. Without arguments and with return type
        3. With arguments and without return type
        4. With arguments and with return type
Arbitrary Arguments
- We can pass mutiple arguments by adding * in the paramter arguments(*s)
- Output in tuple
Keyword Arguments
- if we pass key value paried(dict) in arugments we use **(**s).Result dict
-Default Parameters - (a=None or a=3)

```python
#Without arguments and without return type
def greet():
    print("Hello World")
greet()
#Without arguments and with return type
def number():
    return 5
print(number())
#With arguments and without return type
def sum1(a,b): #Parameteres
    print(a+b)
result = sum1(2,9) #Arguments
#With arguments and with return type
def mul(a,b):
    return a*b
print(mul(2,3))
```

```python
#Arbitrary Arguments
def scores(*s):
    print(s)
scores(25,25,26)
#Keyword Arguments
def my_funcation(**s):
    print(s)
my_funcation(name="hari",age=10,rollnumber=5)

def mail(**details):
    for key,value in details.items():
        print(f"{key.capitalize()}:{value}")
mail(to="hari@gmail.com",subject="pythonclass",body="class at 7pm")

```
