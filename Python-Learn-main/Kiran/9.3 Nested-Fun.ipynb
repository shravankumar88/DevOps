>Nested Function

>MAP
1. Map function takes a function and one or more iterables as input,applies the function to each corresponding element in the iterables, and returns an iterator of the result
2. Syntax: map(function,iterable..)
>Filter
1. Used to filter elements in iterable, based on a specfic function for which the provideed funtion returns TRUE
2. Syntax : filter(function,sequence)
3. funcion: Tests if each elment of a sequence true or not
4. Sequence: Which needs to be filtered,it can be sets,lists,tuples or containers of any iterators
5. Returns: An iterator that is already filtered
>Reduce
1. Builtfunction in python functools module.
2. As the name desceibes applies a given function to the iterables and retuns a single value
3. Takes two arguments: to perfrom the computation and the iterable to be reduced
4. Syntax: from functools import reduce: reduce (function,sequence)
>Lambda
- lambda is a small anonymous function
- Can take any number of arguments, but can only have one expression
- syntax: lambda argumnets: expression
- f = lambda a: a* a
- f function object/name that accepts and stores the result of the expression
- lambda is keyword
- a is argument
- a*a one-line expression
>Genearator Function
1. in funcation if yield contains it is generator function and it controls the flow of the function
2. It pauses the function execution by saving all ststaes and later resume execution when a successvie function is called
  

```python
#Nested Function
def outer_function():
    print("This is outer function")
    def inner_function():
        print("This is inner function")
    inner_function()
outer_function()
```

```python
def calclulate():
    def add(a,b):
        return a+b
    result= add(2,4)
    print(result)
calclulate()
```

```python
def greet(name):
    def wish(n):
        return n.title()
    return f"Hello {wish(name)}"
print(greet("suresh"))

def message(msg):
    def inner():
        return msg*4
    return inner()
print(message("Hello"))
```

```python
#lambda
#can take any num of arrguments but only one expression
#lambda arugments:expression
f = lambda a,b,c:a+b+c
print(f(2,3,4))

numbers = [2,4,5,6]
numbers1 =[]
for i  in numbers:
    f = lambda a:a+1
    numbers1.append(f(i))
print(numbers1)

numbers3 = [2,6,4,5,8,9,10]
evn_num1 = filter(lambda n:n%2==0,numbers3)
print(list(evn_num1))
```

```python
#FIlter Function
ages=[8,21,0,45,13]
def voter_check(x):
    if x>18:
        return True
    else:
        return False
voters=filter(voter_check,ages)
print(list(voters))

numbers2 = [2,6,4,5,8,9,10]
def is_even(n):
    return n %2 ==0
evn_num=filter(is_even,numbers2)
print(list(evn_num))

numbers3 = [2,6,4,5,8,9,10]
evn_num1 = filter(lambda n:n%2==0,numbers3)
print(list(evn_num1))

words=["dog","cat","cow","peackcock"]
filter_words =filter(lambda words:words.startswith("c"),words)
print(list(filter_words))
```

```python
#Map function
#map(function,ietrable...)
def square(n):
    return n*n
numbers4=[2,4,6,7]
result = map(square,numbers4)
print(list(result))

list1 = [2,3,4,5]
list2 = [6,7,8,9]
def my_function(a,b):
    return a+b
result1= map(my_function,list1,list2)
print(list(result1))
```

```python
#reduce
from functools import reduce
numbers5 =[2,4,6,8]
result=reduce(lambda a,b:a*b,numbers5)
print(result)
```

```python
def sum(a,b):
    return a+b
numbers6 = [1,2,3,4]
result=reduce(sum,numbers6)
print(result)

numbers7=[3,7,8,2,5]
value=reduce(lambda a,b:a if a>b else b, numbers7)
print(value)
```

```python
#Generator function
def simplegeneratorfun():
    yield 5
    yield 3
x=simplegeneratorfun()
print(next(x))
print(next(x))

for i in x:
    print(i)
    
```
