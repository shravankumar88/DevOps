```python
#75. From a list like ["Mahesh", "pavan123", "suresh", "ch3rry"],
# print only the words that are purely alphabetic.
purely_alphabetic = ["Mahesh", "pavan123", "suresh", "ch3rry"]
for i in purely_alphabetic:
    if i.isalpha():
        print(i)
```

```python
#76. Ask for a username. Accept only if it's alphanumeric using isalnum().
contains_alphanumeric = input("Enter the name with numbers")
for i in contains_alphanumeric:
    if i == i.isalnum():
        print(f"{i}, Yes it is alphanumeric ")
    else:
        print(f"{i}, Yes it is not alphanumeric ")
```
