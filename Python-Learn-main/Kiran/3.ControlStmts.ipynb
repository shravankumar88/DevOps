### Contorl Stmts
- Conditional Stmt(Decision Making) = 
    - if, if - else, if - elif- else, nested if - else, shorthand if
    - We use those stmts while we want to execite a block of code when the given condition is True or False
    - Shorthand if else can be used for one-lner ternary operations
        - x if condition elase y
            - if condition is True, it returns x
            - if condition is False, it returns y
- Transfer Stmt (Jumping stmts) 
    - break, contuine, pass
- Iterative Stmt (Looping Stmts) = for, while
    - While Loop: Repeats the stmt or group of stmts while a given condition is TRUE. It tests the condition before executing the loop body.
    - For Loop: Executes the code block mutiple times and abbreviates the code that manage the loop variable
        - range(stop), range(start,stop),range(start,stop,skip-1)
    - Nested Loop: We can iterate a loop inside another loop

            - break 
- continue: SKip the false condition
    - for item in iterable:
        - if condition:
            - continue 
- Pass: Is a null operation

```python
for i in range(1,20):
    if i ==5:
        break
    print(i)
```
