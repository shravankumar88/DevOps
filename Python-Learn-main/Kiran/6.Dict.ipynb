```python
#pop
my_dictionary3={
    'user1': 'sai',
 'user2': 'mahesh',
 'user3': 'chiatnya',
 'user4': 'updating the new user',
 'user5': 'addnewuser'
}
my_dictionary3.pop(("user1"))
print(my_dictionary3)
#update
my_dictionary3.update({"user1" :"addnewuser1"}) #adding new user
print(my_dictionary3)
```

```python
#differnt FOR Loops
my_dictionary4={
    'user1': 'sai',
 'user2': 'mahesh',
 'user3': 'chiatnya',
 'user4': 'updating the new user',
 'user5': 'addnewuser'
}
for i in my_dictionary4.items():
    print(i)
for (i,j) in my_dictionary4.items():
    print(i,j)
for i in my_dictionary4:
    print(my_dictionary4[i])
for i in my_dictionary4:
    print(my_dictionary4.get(i))
```

```python
#85. Given a list of fruit names 
# ["apple", "banana", "apple", "orange", "banana", "apple"], 
# count how many of each fruit is in the basket.
fruits_name = ["apple", "banana", "apple", "orange", "banana", "apple"]
fruit_dict = {}

for fruit in fruits_name:
    fruit_dict[fruit] = fruit_dict.get(fruit, 0) + 1

print(fruit_dict)
```
