```python
balance = 1000 
def withdraw(amount):
    global balance # need to define the global to pass the balance
    if amount<=balance:
        balance-=amount
        print(f"withdraw {amount}, New balnce:{balance}")
    else:
        print("insifficent funds")
withdraw(200)
```

```python
def withdraw(balance,amount): # passing the balance in paramater
    if amount<=balance:
        balance-=amount
        print(f"withdraw {amount}, New balnce:{balance}")
    else:
        print("insifficent funds")
withdraw(1000,200)
```
