A set is a collection which is unordered and unindexed.
- Set items are unordered,and do not allow duplicate values.
- Unordered means that the items in a set do not have a defined order.
-- Set items can appear in a different order every time you use them,and
cannot be referred to by index or key.
- A set is created by placing all the items (elements) inside curly braces
{},separated by comma,or by using the built-in set() function.
- It can have any number of items and they may be of different types
(integer,float,tuple,string,etc).But a set cannot have mutable elementals
like lists,sets or dictionaries as its elements


```python
#100. Create two sets of your chocic.
#a. union,intersection, differnence of both sets(a-B &B-a), symmetric differnece
set_1 = {1,2,3,4,5}
set_2 = {2,5,6,7,8,9}
set_union = set_1.union(set_2)
print(f"set_union:{set_union}")
set_intersection = (set_1.intersection(set_2))
print(f"set_intersection: {set_intersection}")
set_diffA_B = set_1.difference(set_2)
print(f"set_diffA_B: {set_diffA_B}")
set_diffB_A = set_2.difference(set_1)
print(f"set_diffB_A: {set_diffB_A}")
set_symmetric = set_1.symmetric_difference(set_2)
print(f"set_symmetric:{set_symmetric}")
```

```python
#101. Create two sets  one being a subset of the other
# a.Check and print whether the first set is a subset of the second
# b. Whether the second set us a superset of the first
# c. check membership of an elemnt that is not in either set
set_3 = {5,6,7,8,9}
set_4 = {7,8,10,1,2}
set_issubset = set_3.issubset(set_4)
print(f"set_issubset:{set_issubset}")
set_issuperset = set_3.issuperset(set_4)
print(f"set_issuperset:{set_issuperset}")
```
