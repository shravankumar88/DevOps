>Encapsulation (Binding of the class methods and attrbuties)
1. Is a mechanism of wrapping the data/attributes(variables) and code acting on the data(methods) together as a single unit.
2. Access Modifiers
    - Private   -   Not Acessible outside from class
        - Double underscore(__)
        - Can`t be accessed directly from outside the class
        - Used to hide sensitive data
    - Protected -   Accessible to derived/child class
        - One underscore(_)
        - Meant to be used within the class and subclasses
    - Public    -   Accessible to Outside world
        - NO Underscore
        - can be accessed from anywhere
>Abstraction
1. That allows you to hide the complex details of a system and expose the essential Features
2. It focuses on what an object does, rather than how it does it.


```python
#Encapsulation (Binding of the class methods and attrbuties)
#Private
class Grandfather():
    def __init__(self,a):
        self.__y =a #__ which is private
        print(self.__y)
class Father(Grandfather):
    def display(self):
        print("Father Class",self.__y)
class Child(Father):
    def output(self):
        print("Child Class",self.__y)
c=Child(30)
c.display()
c.output()
```

```python
#protected
class Grandfather():
    def __init__(self,a):
        self._y =a #__ which is protected
        print(self._y)
class Father(Grandfather):
    def display(self):
        print("Father Class",self._y)
class Child(Father):
    def output(self):
        print("Child Class",self._y)
c=Child(30)
c.display()
c.output()
```
