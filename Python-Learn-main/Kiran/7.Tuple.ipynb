```python
# 96. Create two lists: one of names and one of ages.
# Use zip() to pair each name with the corresponding age into a tuple.
names1 = ['str', 2, 3.4, "str1"]
age = (20, 30, 40, 50)
result1 = zip(names1, age)
print(list(result1))
```

```python
# 97. Zip three lists together and display the result.
names2 = ['str', 2, 3.4, "str1"]
age1 = (20, 30, 40, 50)
rollnumber = (2.3,4.5,6.7,8.9)
result2 = zip(names2,age1,rollnumber)
print(tuple(result2))
result3 = (('str', 20, 2.3), (2, 30, 4.5), (3.4, 40, 6.7), ('str1', 50, 8.9))
names4,age2,rollnumber1 = zip(*result3) #unzip
print(names4,age2,rollnumber1)
```

```python
# *
# **
# ***
# ****
# *****
# ******
for i in range(7):
    if i == i:
        print ("*" * i)
    else:
        print("it is done")
```

```python
for i in range(5):
    for i in range(i+1):
        print("*", end="")
    print()
```

```python
for i in range(5,0,-1):
    for j in range(i):
        print("*", end="")
    print()
```
