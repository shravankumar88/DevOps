>Inheritance 
1. Allows us to define a class that inherits all the methods and properties from another class
2. Parent Class is the class being inhertied from,also called base/super class
3. Child Class is the class that inherits from another class, also called derived/sub class
>Polymorphism
1. Poly - Many Morph - forms
2. Refers to the ability of differnt class to define methods with the same name but with differnt implemntataions.
3. It allows objects of different types to be treated as instances of the same class through a common interface, enabaling flexibility and reuse in code.
4. Types of polymorphism
    > Compile Time Polymorphism (Method Overloading)
    1. When mutiple methods have the same name but differnt parameteres
    2. Python doesn`t support this,it can be simulted using default arguments
    > Run-Time Polymorphism     (Method Overriding)
    1. When a subclass provides a specgic implementation of a method is already defined in a parent class
    2. The appropriate method is chosen based on the object`s type at runtime
>super() function
1. Allows the parent class`s attributes and methods to be fully accessible within a subclass
2. It extends the functionality of the superclass within the subclass or derived class

```python
#Single Inheritance(Parent,child)
class Parent():
    def output(self):
        print("This is Parent class")
class Child(Parent): #Inherit the Child from Parent
    def outputchild(self):
        print("This is child class")
Child_obj=Child()
Child_obj.output()
```

```python
#Mutiple Inheritance(Mutiple base class and single child class)
class Father():
    def outputfather(self):
        print("This is Father class")
class Mother():
    def outputmother(self):
        print("This is Mother class")
class Child(Father,Mother):
    def outputchild(self):
        print("This is Child class")                
Child_obj1=Child()
Child_obj1.outputchild()
Child_obj1.outputmother()
Child_obj1.outputfather()
```

```python
#MultiLevel Inheritance(Level by level)
class GrandFather():
    def grandFather(self):
        print("This is GrandFather Class")
class Father(GrandFather):
    def father(self):
        print("This is Father class")
class Child1(Father):
    def child(self):
        print("This is child class")
Child_obj2 = Child1()
Child_obj2.child()
Child_obj2.father()
Child_obj2.grandFather()             
```

```python
#HierarchicalInheritnace(One parent class and multiple child class)
class Parent1():
    def parent1(self):
        print("This is parent1 class")
class child1(Parent1):
    def child1(self):
        print("This is child1 class")
class child2(Parent1):
    def child2(self):
        print("This is child2 class")
child1_obj = child1()
child1_obj.child1()
child1_obj.parent1()
child3_obj = child2()
child3_obj.child2()
child3_obj.parent1()
```

```python
#Hybrid Inheritance (Combination of more than one type of a inheritance in a single program)
class Animal():
    def speaks(self):
        print("This is Animal class")
class dog(Animal): #single Inheritance
    def braks(self):
        print("This is Dog class")
class Pet():
    def is_pet(self):
        print("This is Pet class")
class PetDog(dog,Pet): #Multiple Inheritance
    def is_petdog(self):
        print("This is PetDog class")   
child_dog=dog()
child_dog.braks()
child_dog.speaks()
child_petdog=PetDog()
child_petdog.is_petdog()
child_petdog.is_pet()
child_petdog.braks()
child_petdog.speaks()             
```

```python
#Method overriding 
#Both class have same function, when we call the object CHILD class with print
class Animal(): #parent class
    def speaks(self):
        print("This is Animal class")
class dog(Animal): #child class
    def speaks(self):
        print("This is Dog class")
object3=dog()
object3.speaks()
# how to access the Parent Function
#using Suepr() Function - To access the Parent class Method in child class
```

```python
#using Suepr() Function 
class Animal(): #parent class
    def speaks(self):
        print("This is Animal class")
class dog(Animal): #child class
    def speaks(self):
        super().speaks()
        print("This is Dog class")
object3=dog()
object3.speaks()
```

```python
# 133. Write a function that accepts any object that has a method area().
# Create classes Circle, Rectangle, and Square 
# that implement area() and pass them to the function.
import math

class Circle:
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2

class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

class Square:
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side ** 2

# Duck typing function
def print_area(shape):
    print(f"Area: {shape.area()}")

# Create objects
circle = Circle(5)
rectangle = Rectangle(4, 6)
square = Square(3)

# Call function with different shapes
print_area(circle)     # Area of Circle
print_area(rectangle)  # Area of Rectangle
print_area(square)     # Area of Square

```
