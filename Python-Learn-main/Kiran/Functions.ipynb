Function
- A block of code whichruns when it is called.
- parameters --> Function Defintion --> Formal Parameters
- Arguments  --> Function Calling   --> Actual Parameters
- Return is used to exit the function and return  a value to the caller

```python
def sum():
    number1=int(input("Enter first number:"))
    number2=int(input("Enter second number:"))
    result = number1+number2
    print(result)
sum()#Function calling
```

```python
def sum(a,b): #Parameteres
    print(a+b)
sum(2,3) #Arguments
def sum1(a,b): #Parameteres
    print(a+b)
    return a+b
sum1(2,9) #Arguments
```
