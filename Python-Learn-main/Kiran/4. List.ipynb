| Data Structure | What It Allows                                                                                                                                                                       |
| -------------- | ------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------ |
| **List**       | Any data type: `int`, `float`, `str`, `bool`, `list`, `tuple`, `dict`, `set`, `None`, custom objects. Example: `[1, "apple", 3.14, True, [5,6], {"a":1}]`                            |
| **Set**        | Any **immutable** type: `int`, `float`, `str`, `tuple` (if elements inside the tuple are also immutable). ❌ Cannot include lists or dicts. Example: `{1, "apple", (2, 3)}`           |
| **Tuple**      | Any data type, including mutable ones like `list`, `dict`. Example: `(1, "apple", [5,6], {"a":1})`                                                                                   |
| **Dict**       | **Keys:** Must be immutable (`str`, `int`, `float`, `tuple` with immutable contents). <br> **Values:** Can be any type. Example: `{"name": "Alice", 42: [1, 2], (1,2): "tuple key"}` |

```python
# 57.Using List Comprehension 
# Create a list of squares of even numbers from 20 to 30
list_square = []
for i in range(20,30):
    list_square.append(i**2)
print(list_square)
###
list_square1=[i**2 for i in range(20,30)]
print(list_square1)
```

```python
# 58.Take a range(10,21) and 
# if the number is even multiply by 2 and if the number is odd then take 0 
# by using List comprehension display the output?
take_zero=[i*2 if i%2==0 else 0 for i in range(10,21)]
print(take_zero)
```
